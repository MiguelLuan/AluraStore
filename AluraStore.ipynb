# 3. Receita por Estado

```python
# Faturamento da loja1
receita_estado_loja1 = loja1.groupby('Local da compra').agg(
    Faturamento = ('Preço','sum'),
    Quantidade = ('Local da compra','size')
)
receita_estado_loja1.sort_values(by='Faturamento',ascending=False,inplace=True)
receita_estado_loja1
```

```python
# Faturamento da loja2
receita_estado_loja2 = loja2.groupby('Local da compra').agg(
    Faturamento = ('Preço','sum'),
    Quantidade = ('Local da compra','size')
)
receita_estado_loja2.sort_values(by='Faturamento',ascending=False,inplace=True)
receita_estado_loja2
```

```python
# Faturamento da loja3
receita_estado_loja3 = loja3.groupby('Local da compra').agg(
    Faturamento = ('Preço','sum'),
    Quantidade = ('Local da compra','size')
)
receita_estado_loja3.sort_values(by='Faturamento',ascending=False,inplace=True)
receita_estado_loja3
```

# 4. Vendas Por Categoria

```python
# Quantidade de produtos vendidos junto ao faturamento da loja1
qtd_produtos_loja1 = loja1.groupby('Categoria do Produto').agg(
    Quantidade_de_Produtos = ('Categoria do Produto','size'),
    Preco_Total = ('Preço','sum'),
    Frete_Total = ('Frete','sum'),
)

qtd_produtos_loja1['Faturamento'] = qtd_produtos_loja1['Preco_Total'] - qtd_produtos_loja1['Frete_Total']

qtd_produtos_loja1.drop(columns=['Preco_Total','Frete_Total'],inplace=True)

qtd_produtos_loja1
```

```python
# Quantidade de produtos vendidos junto ao faturamento da loja2
qtd_produtos_loja2 = loja2.groupby('Categoria do Produto').agg(
    Quantidade_de_Produtos = ('Categoria do Produto','size'),
    Preco_Total = ('Preço','sum'),
    Frete_Total = ('Frete','sum'),
)

qtd_produtos_loja2['Faturamento'] = qtd_produtos_loja2['Preco_Total'] - qtd_produtos_loja2['Frete_Total']

qtd_produtos_loja2.drop(columns=['Preco_Total','Frete_Total'],inplace=True)

qtd_produtos_loja2
```

```python
# Quantidade de produtos vendidos junto ao faturamento da loja3
qtd_produtos_loja3 = loja3.groupby('Categoria do Produto').agg(
    Quantidade_de_Produtos = ('Categoria do Produto','size'),
    Preco_Total = ('Preço','sum'),
    Frete_Total = ('Frete','sum'),
)

qtd_produtos_loja3['Faturamento'] = qtd_produtos_loja3['Preco_Total'] - qtd_produtos_loja3['Frete_Total']

qtd_produtos_loja3.drop(columns=['Preco_Total','Frete_Total'],inplace=True)

qtd_produtos_loja3
```

```python
# Quantidade de produtos vendidos junto ao faturamento da loja4
qtd_produtos_loja4 = loja4.groupby('Categoria do Produto').agg(
    Quantidade_de_Produtos = ('Categoria do Produto','size'),
    Preco_Total = ('Preço','sum'),
    Frete_Total = ('Frete','sum'),
)

qtd_produtos_loja4['Faturamento'] = qtd_produtos_loja4['Preco_Total'] - qtd_produtos_loja4['Frete_Total']

qtd_produtos_loja4.drop(columns=['Preco_Total','Frete_Total'],inplace=True)

qtd_produtos_loja4
```




# 5. Produtos Mais e Menos Vendidos


```python
# Produtos mais e menos vendidos da loja1
pdt_vdd_loja1 = loja1.groupby(['Produto', 'Categoria do Produto']).agg(
    Quantidade_de_Produtos=('Produto', 'size'),
    Preco=('Preço', 'mean')
)

pdt_mais_vdd_loja1 = pdt_vdd_loja1['Quantidade_de_Produtos'].idxmax()
pdt_menos_vdd_loja1 = pdt_vdd_loja1['Quantidade_de_Produtos'].idxmin()

produto_max_loja1, categoria_max_loja1 = pdt_mais_vdd_loja1
produto_min_loja1, categoria_min_loja1 = pdt_menos_vdd_loja1

quantidade_pdt_max_loja1 = int(pdt_vdd_loja1.loc[pdt_mais_vdd_loja1, 'Quantidade_de_Produtos'])
quantidade_pdt_min_loja1 = int(pdt_vdd_loja1.loc[pdt_menos_vdd_loja1, 'Quantidade_de_Produtos'])

preco_pdt_max_loja1 = float(pdt_vdd_loja1.loc[pdt_mais_vdd_loja1, 'Preco'])
preco_pdt_min_loja1 = float(pdt_vdd_loja1.loc[pdt_menos_vdd_loja1, 'Preco'])

print(f'Produto mais vendido: {produto_max_loja1} ({categoria_max_loja1})')
print(f'Quantidade: {quantidade_pdt_max_loja1}, Preço médio: R${preco_pdt_max_loja1:.2f}\n')

print(f'Produto menos vendido: {produto_min_loja1} ({categoria_min_loja1})')
print(f'Quantidade: {quantidade_pdt_min_loja1}, Preço médio: R${preco_pdt_min_loja1:.2f}')
```
