# 3. Vendas Por Categoria

```python
# Quantidade de produtos vendidos junto ao faturamento da loja1
qtd_produtos_loja1 = loja1.groupby('Categoria do Produto').agg(
    Quantidade_de_Produtos = ('Categoria do Produto','size'),
    Preco_Total = ('Preço','sum'),
    Frete_Total = ('Frete','sum'),
)

qtd_produtos_loja1['Faturamento'] = qtd_produtos_loja1['Preco_Total'] + qtd_produtos_loja1['Frete_Total']

qtd_produtos_loja1.drop(columns=['Preco_Total','Frete_Total'],inplace=True)

qtd_produtos_loja1
```

```python
# Quantidade de produtos vendidos junto ao faturamento da loja2
qtd_produtos_loja2 = loja2.groupby('Categoria do Produto').agg(
    Quantidade_de_Produtos = ('Categoria do Produto','size'),
    Preco_Total = ('Preço','sum'),
    Frete_Total = ('Frete','sum'),
)

qtd_produtos_loja2['Faturamento'] = qtd_produtos_loja2['Preco_Total'] + qtd_produtos_loja2['Frete_Total']

qtd_produtos_loja2.drop(columns=['Preco_Total','Frete_Total'],inplace=True)

qtd_produtos_loja2
```

```python
# Quantidade de produtos vendidos junto ao faturamento da loja3
qtd_produtos_loja3 = loja3.groupby('Categoria do Produto').agg(
    Quantidade_de_Produtos = ('Categoria do Produto','size'),
    Preco_Total = ('Preço','sum'),
    Frete_Total = ('Frete','sum'),
)

qtd_produtos_loja3['Faturamento'] = qtd_produtos_loja3['Preco_Total'] + qtd_produtos_loja3['Frete_Total']

qtd_produtos_loja3.drop(columns=['Preco_Total','Frete_Total'],inplace=True)

qtd_produtos_loja3
```

```python
# Quantidade de produtos vendidos junto ao faturamento da loja4
qtd_produtos_loja4 = loja1.groupby('Categoria do Produto').agg(
    Quantidade_de_Produtos = ('Categoria do Produto','size'),
    Preco_Total = ('Preço','sum'),
    Frete_Total = ('Frete','sum'),
)

qtd_produtos_loja4['Faturamento'] = qtd_produtos_loja4['Preco_Total'] + qtd_produtos_loja4['Frete_Total']

qtd_produtos_loja4.drop(columns=['Preco_Total','Frete_Total'],inplace=True)

qtd_produtos_loja4
```
